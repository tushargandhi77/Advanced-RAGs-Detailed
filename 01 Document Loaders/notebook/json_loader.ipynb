```python
from langchain_community.document_loaders.json_loader import JSONLoader
from pathlib import Path
from pprint import pp
```

```python
file_path = Path("../Knowledge Source/apparels.json")

file_path.exists()
```

```python
def metadata_func(record: dict,metadata: dict) -> dict:
    metadata['product_name'] = record.get("productName")
    metadata['price'] = record.get("price")
    metadata['category'] = record.get("category")
    # del seq number
    del metadata['seq_num']
    return metadata
```

```python
loader = JSONLoader(file_path=file_path.as_posix(),jq_schema=".products[]",content_key="Description",metadata_func=metadata_func)
```

```python
document = loader.load()
```
