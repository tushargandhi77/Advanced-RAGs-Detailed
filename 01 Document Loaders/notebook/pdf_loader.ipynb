```python
from langchain_community.document_loaders.pdf import PyPDFLoader
from pathlib import Path
from pprint import pp
```

```python
file_path = Path("../Knowledge Source/attention_is_all_you_need.pdf")
```

```python
file_path.exists()
```

## PyPDF Loader

```python
file_path.as_posix()
```

```python
pypdf_loader = PyPDFLoader(file_path.as_posix(),mode='page')
```

```python
documents = pypdf_loader.load()
```

```python
len(documents)
```

```python
# fetch the text of first page

pp(documents[0].page_content)
```

```python
pp(documents[0].metadata)
```

```python
from langchain_community.document_loaders.parsers import RapidOCRBlobParser
```

```python
# create a pypdf instance that can load images

pypdf_image_loader = PyPDFLoader(file_path.as_posix(),mode='page',extract_images=True,images_parser=RapidOCRBlobParser(),images_inner_format='html-img')
```

```python
documents_with_images = pypdf_image_loader.load()
```

## PDF Miner

```python
from langchain_community.document_loaders import PDFMinerLoader
```

```python
pdfminer_loader = PDFMinerLoader(file_path.as_posix(),mode='page',extract_images=True,images_parser=RapidOCRBlobParser(),images_inner_format='html-img')
```

```python
loaded_documents = pdfminer_loader.load()
```

```python
print(loaded_documents[2].page_content)
```

```python
print(loaded_documents[5].page_content)
```

```python
for doc in loaded_documents:
    if doc.metadata['page'] == 7:
        print(doc.page_content)
```
