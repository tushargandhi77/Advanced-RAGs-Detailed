```python
# create the loaders

loader = PyPDFLoader(file_path='../Openclaw_Research_Report.pdf')

docs = loader.load()

len(docs)
```

```python
# split documents 

chunker = RecursiveCharacterTextSplitter(chunk_size=300, chunk_overlap=50)


chunks = chunker.split_documents(docs)

len(chunks)
```

```python
chunks = chunks[:20]
```

```python
len(chunks)
```

```python
# embed documents

text_documents = [doc.page_content for doc in chunks]

print(len(text_documents))
```

```python
document_embeddings = embeddings.embed_documents(texts=text_documents)
```
